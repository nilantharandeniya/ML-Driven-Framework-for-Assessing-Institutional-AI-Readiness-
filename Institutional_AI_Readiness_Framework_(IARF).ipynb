```python
# IARF ML MODELS - COMPLETE GOOGLE COLAB NOTEBOOK
# Author: Jude Nilantha Randeniya
# Project: AIES 2026 Conference Submission
# Date: May, 2026

"""
This notebook contains all 3 ML models for the Institutional AI Readiness Framework:
- Model 1: Random Forest Regressor (Readiness Scorer)
- Model 2: K-Means + Random Forest Classifier (Profile Classifier)
- Model 3: Logistic Regression (Success Predictor)

HOW TO USE:
1. Upload your 'iarf_dataset_complete.csv' file
2. Run all cells in order
3. Download the results at the end
"""

# ============================================
# SECTION 1: SETUP AND DATA LOADING
# ============================================

print("="*70)
print("IARF ML MODELS - SETUP")
print("="*70)

# Install required packages
!pip install -q scikit-learn pandas numpy matplotlib seaborn

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.cluster import KMeans
from sklearn.metrics import (mean_absolute_error, r2_score, accuracy_score,
                             precision_score, recall_score, f1_score,
                             roc_auc_score, silhouette_score, classification_report,
                             confusion_matrix)
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Set random seed for reproducibility
np.random.seed(42)

print("\n✅ Libraries imported successfully!")

# ============================================
# UPLOAD YOUR DATA
# ============================================

print("\n" + "="*70)
print("STEP 1: UPLOAD YOUR DATA")
print("="*70)
print("\nPlease upload 'iarf_dataset_complete.csv'")

from google.colab import files
uploaded = files.upload()

# Load the dataset
df = pd.read_csv('iarf_dataset_complete.csv')

print(f"\n✅ Data loaded successfully!")
print(f"   - Total organizations: {len(df)}")
print(f"   - Countries: {df['country'].nunique()}")
print(f"   - Sectors: {df['sector'].nunique()}")

# Display first few rows
print("\n📊 Dataset Preview:")
display(df.head())

# Display summary statistics
print("\n📈 Summary Statistics:")
display(df.describe())

# ============================================
# SECTION 2: DATA PREPARATION
# ============================================

print("\n" + "="*70)
print("SECTION 2: DATA PREPARATION")
print("="*70)

# Define dimension columns
dimension_cols = ['gov_maturity', 'tech_capacity', 'transparency',
                  'accountability', 'cultural']

# Create readiness score (0-100)
df['readiness_score'] = df[dimension_cols].mean(axis=1) * 10

# Create binary outcome for success prediction
df['success_binary'] = df['outcome'].apply(
    lambda x: 1 if x in ['Success', 'Partial'] else 0
)

print(f"\n✅ Data prepared!")
print(f"\nReadiness Score Range: {df['readiness_score'].min():.1f} - {df['readiness_score'].max():.1f}")
print(f"Mean Readiness: {df['readiness_score'].mean():.1f}")
print(f"\nOutcome Distribution:")
print(df['outcome'].value_counts())
print(f"\nSuccess Rate: {df['success_binary'].mean()*100:.1f}%")

# ============================================
# SECTION 3: MODEL 1 - READINESS SCORER
# ============================================

print("\n" + "="*70)
print("MODEL 1: READINESS SCORER (Random Forest Regression)")
print("="*70)

# Prepare data
X_model1 = df[dimension_cols]
y_model1 = df['readiness_score']

# Train-test split
X_train_m1, X_test_m1, y_train_m1, y_test_m1 = train_test_split(
    X_model1, y_model1, test_size=0.2, random_state=42
)

print(f"\n📊 Data Split:")
print(f"   Training: {len(X_train_m1)} organizations")
print(f"   Test: {len(X_test_m1)} organizations")

# Train Random Forest Regressor
print(f"\n🔨 Training Random Forest Regressor...")
rf_regressor = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_leaf=5,
    random_state=42
)
rf_regressor.fit(X_train_m1, y_train_m1)

# Predictions
y_pred_m1 = rf_regressor.predict(X_test_m1)

# Evaluation
mae_m1 = mean_absolute_error(y_test_m1, y_pred_m1)
r2_m1 = r2_score(y_test_m1, y_pred_m1)

# Cross-validation
cv_scores_m1 = cross_val_score(rf_regressor, X_model1, y_model1,
                                cv=5, scoring='neg_mean_absolute_error')
cv_mae_m1 = -cv_scores_m1.mean()

print(f"\n✅ MODEL 1 RESULTS:")
print(f"   MAE: {mae_m1:.2f} points")
print(f"   R²: {r2_m1:.3f} ({r2_m1*100:.1f}% variance explained)")
print(f"   Cross-validated MAE: {cv_mae_m1:.2f}")

# Feature importance
feature_importance_m1 = pd.DataFrame({
    'dimension': dimension_cols,
    'importance': rf_regressor.feature_importances_,
    'importance_pct': rf_regressor.feature_importances_ * 100
}).sort_values('importance', ascending=False)

print(f"\n📈 FEATURE IMPORTANCE:")
for idx, row in feature_importance_m1.iterrows():
    print(f"   {row['dimension']:20s}: {row['importance_pct']:5.1f}%")

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Actual vs Predicted
ax1.scatter(y_test_m1, y_pred_m1, alpha=0.6, s=100, edgecolors='black')
ax1.plot([y_model1.min(), y_model1.max()],
         [y_model1.min(), y_model1.max()], 'r--', lw=2)
ax1.set_xlabel('Actual Readiness', fontsize=12, fontweight='bold')
ax1.set_ylabel('Predicted Readiness', fontsize=12, fontweight='bold')
ax1.set_title('Model 1: Actual vs Predicted', fontsize=14, fontweight='bold')
ax1.text(0.05, 0.95, f'R² = {r2_m1:.3f}\nMAE = {mae_m1:.2f}',
         transform=ax1.transAxes, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
ax1.grid(True, alpha=0.3)

# Feature Importance
ax2.barh(feature_importance_m1['dimension'],
         feature_importance_m1['importance_pct'],
         color='steelblue', edgecolor='black')
ax2.set_xlabel('Importance (%)', fontsize=12, fontweight='bold')
ax2.set_title('Feature Importance', fontsize=14, fontweight='bold')
ax2.grid(True, axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('model1_performance.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✅ Model 1 Complete!")

# ============================================
# SECTION 4: MODEL 2 - PROFILE CLASSIFIER
# ============================================

print("\n" + "="*70)
print("MODEL 2: PROFILE CLASSIFIER (K-Means + Random Forest)")
print("="*70)

# Standardize features for K-means
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[dimension_cols])

# K-means clustering (k=5 for 5 profiles)
print(f"\n🔨 Running K-Means clustering (k=5)...")
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df['profile'] = kmeans.fit_predict(X_scaled)

# Map to letters A-E
profile_mapping = {i: chr(65+i) for i in range(5)}
df['profile_label'] = df['profile'].map(profile_mapping)

# Silhouette score
silhouette = silhouette_score(X_scaled, df['profile'])

print(f"\n✅ Clustering complete!")
print(f"   Silhouette Score: {silhouette:.3f}")

# Profile characteristics
print(f"\n📊 PROFILE CHARACTERISTICS:")
profile_summary = df.groupby('profile_label')[dimension_cols + ['readiness_score', 'success_binary']].mean()
display(profile_summary.round(2))

print(f"\n📊 PROFILE DISTRIBUTION:")
print(df['profile_label'].value_counts().sort_index())

# Train Random Forest Classifier for profile prediction
print(f"\n🔨 Training Profile Classifier...")
X_train_m2, X_test_m2, y_train_m2, y_test_m2 = train_test_split(
    df[dimension_cols], df['profile'], test_size=0.2, random_state=42
)

rf_classifier = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42
)
rf_classifier.fit(X_train_m2, y_train_m2)

# Evaluate
y_pred_m2 = rf_classifier.predict(X_test_m2)
accuracy_m2 = accuracy_score(y_test_m2, y_pred_m2)

print(f"\n✅ MODEL 2 RESULTS:")
print(f"   Silhouette Score: {silhouette:.3f}")
print(f"   Classification Accuracy: {accuracy_m2:.1%}")

# Visualization - Profile Radar Chart
fig, ax = plt.subplots(figsize=(10, 8), subplot_kw=dict(projection='polar'))

# Prepare data for radar chart
categories = ['Governance', 'Technical', 'Transparency', 'Accountability', 'Cultural']
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
angles += angles[:1]  # Complete the circle

# Plot each profile
colors = ['red', 'blue', 'green', 'orange', 'purple']
for i, (profile, color) in enumerate(zip(['A', 'B', 'C', 'D', 'E'], colors)):
    if profile in profile_summary.index:
        values = profile_summary.loc[profile, dimension_cols].tolist()
        values += values[:1]  # Complete the circle
        ax.plot(angles, values, 'o-', linewidth=2, label=f'Profile {profile}', color=color)
        ax.fill(angles, values, alpha=0.15, color=color)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=11)
ax.set_ylim(0, 10)
ax.set_title('Profile Characteristics (Radar Chart)',
             fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
ax.grid(True)

plt.tight_layout()
plt.savefig('model2_profiles.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✅ Model 2 Complete!")

# ============================================
# SECTION 5: MODEL 3 - SUCCESS PREDICTOR
# ============================================

print("\n" + "="*70)
print("MODEL 3: SUCCESS PREDICTOR (Logistic Regression)")
print("="*70)

# Prepare data
X_model3 = df[['readiness_score'] + dimension_cols]
y_model3 = df['success_binary']

# Train-test split
X_train_m3, X_test_m3, y_train_m3, y_test_m3 = train_test_split(
    X_model3, y_model3, test_size=0.2, random_state=42
)

print(f"\n📊 Data Split:")
print(f"   Training: {len(X_train_m3)} organizations")
print(f"   Test: {len(X_test_m3)} organizations")

# Train Logistic Regression
print(f"\n🔨 Training Logistic Regression...")
log_reg = LogisticRegression(random_state=42, max_iter=1000)
log_reg.fit(X_train_m3, y_train_m3)

# Predictions
y_pred_m3 = log_reg.predict(X_test_m3)
y_pred_proba_m3 = log_reg.predict_proba(X_test_m3)[:, 1]

# Evaluation
accuracy_m3 = accuracy_score(y_test_m3, y_pred_m3)
precision_m3 = precision_score(y_test_m3, y_pred_m3, zero_division=0)
recall_m3 = recall_score(y_test_m3, y_pred_m3, zero_division=0)
f1_m3 = f1_score(y_test_m3, y_pred_m3, zero_division=0)

# Handle AUC calculation (need at least 2 classes)
if len(np.unique(y_test_m3)) > 1:
    auc_m3 = roc_auc_score(y_test_m3, y_pred_proba_m3)
else:
    auc_m3 = 0.0

print(f"\n✅ MODEL 3 RESULTS:")
print(f"   Accuracy: {accuracy_m3:.1%}")
print(f"   Precision: {precision_m3:.2f}")
print(f"   Recall: {recall_m3:.2f}")
print(f"   F1 Score: {f1_m3:.2f}")
print(f"   AUC: {auc_m3:.2f}")

# Coefficients
coef_df = pd.DataFrame({
    'feature': X_model3.columns,
    'coefficient': log_reg.coef_[0]
}).sort_values('coefficient', key=abs, ascending=False)

print(f"\n📈 LOGISTIC REGRESSION COEFFICIENTS:")
for idx, row in coef_df.iterrows():
    print(f"   {row['feature']:20s}: {row['coefficient']:7.3f}")

# Find governance vs technical coefficients
gov_coef = coef_df[coef_df['feature'] == 'gov_maturity']['coefficient'].values[0] if 'gov_maturity' in coef_df['feature'].values else 0
tech_coef = coef_df[coef_df['feature'] == 'tech_capacity']['coefficient'].values[0] if 'tech_capacity' in coef_df['feature'].values else 0

print(f"\n🔑 KEY FINDING:")
print(f"   Governance coefficient: {gov_coef:.3f}")
print(f"   Technical coefficient: {tech_coef:.3f}")

# Confusion Matrix
cm = confusion_matrix(y_test_m3, y_pred_m3)
print(f"\n📊 CONFUSION MATRIX:")
print(cm)

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Confusion Matrix Heatmap
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax1,
            xticklabels=['Failed', 'Success'],
            yticklabels=['Failed', 'Success'])
ax1.set_xlabel('Predicted', fontsize=12, fontweight='bold')
ax1.set_ylabel('Actual', fontsize=12, fontweight='bold')
ax1.set_title('Confusion Matrix', fontsize=14, fontweight='bold')

# Coefficient comparison
coef_plot = coef_df.head(6)  # Top 6 features
colors = ['green' if x > 0 else 'red' for x in coef_plot['coefficient']]
ax2.barh(coef_plot['feature'], coef_plot['coefficient'],
         color=colors, edgecolor='black')
ax2.set_xlabel('Coefficient', fontsize=12, fontweight='bold')
ax2.set_title('Feature Coefficients', fontsize=14, fontweight='bold')
ax2.axvline(x=0, color='black', linestyle='-', linewidth=0.8)
ax2.grid(True, axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('model3_results.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n✅ Model 3 Complete!")

# ============================================
# SECTION 6: KEY FINDINGS FOR PAPER
# ============================================

print("\n" + "="*70)
print("SECTION 6: KEY FINDINGS FOR YOUR PAPER")
print("="*70)

# Implementation Paradox
govt_success = df[df['sector']=='Government']['success_binary'].mean() * 100
acad_success = df[df['sector']=='Academic']['success_binary'].mean() * 100
ngo_success = df[df['sector']=='NGO']['success_binary'].mean() * 100

print(f"\n📊 IMPLEMENTATION PARADOX:")
print(f"   Government adoption: {govt_success:.0f}%")
print(f"   Academic adoption: {acad_success:.0f}%")
print(f"   NGO adoption: {ngo_success:.0f}%")

# Profile success rates
print(f"\n📊 PROFILE SUCCESS RATES:")
for profile in sorted(df['profile_label'].unique()):
    profile_data = df[df['profile_label'] == profile]
    success_rate = profile_data['success_binary'].mean() * 100
    count = len(profile_data)
    print(f"   Profile {profile}: {success_rate:.0f}% ({count} orgs)")

# Critical threshold
threshold_75 = df['readiness_score'] >= 75
above_75 = df[threshold_75]['success_binary'].mean() * 100
below_75 = df[~threshold_75]['success_binary'].mean() * 100

print(f"\n📊 CRITICAL THRESHOLD (75/100):")
print(f"   Success rate above 75: {above_75:.0f}%")
print(f"   Success rate below 75: {below_75:.0f}%")

# ============================================
# SECTION 7: SAVE RESULTS
# ============================================

print("\n" + "="*70)
print("SECTION 7: SAVING RESULTS")
print("="*70)

# Create results summary
results_summary = {
    'Metric': [
        'Sample Size',
        'Model 1 MAE',
        'Model 1 R²',
        'Model 2 Silhouette',
        'Model 2 Accuracy',
        'Model 3 Accuracy',
        'Model 3 AUC',
        'Govt Adoption %',
        'Academic Adoption %',
        'NGO Adoption %'
    ],
    'Value': [
        len(df),
        mae_m1,
        r2_m1,
        silhouette,
        accuracy_m2,
        accuracy_m3,
        auc_m3,
        govt_success,
        acad_success,
        ngo_success
    ]
}

results_df = pd.DataFrame(results_summary)
results_df.to_csv('all_results_summary.csv', index=False)

# Save profile data
df[['org_id', 'profile_label', 'readiness_score', 'outcome']].to_csv('organization_profiles.csv', index=False)

print("\n✅ Results saved!")
print("\nFiles created:")
print("  1. model1_performance.png")
print("  2. model2_profiles.png")
print("  3. model3_results.png")
print("  4. all_results_summary.csv")
print("  5. organization_profiles.csv")

# Download all files
print("\n" + "="*70)
print("DOWNLOAD YOUR RESULTS")
print("="*70)

from google.colab import files
files.download('model1_performance.png')
files.download('model2_profiles.png')
files.download('model3_results.png')
files.download('all_results_summary.csv')
files.download('organization_profiles.csv')

print("\n" + "="*70)
print("✅ ALL MODELS COMPLETE!")
print("="*70)
print(f"""
🎉 SUCCESS! All 3 ML models have been built and validated!

📊 QUICK SUMMARY:
- Model 1 (Readiness Scorer): MAE={mae_m1:.1f}, R²={r2_m1:.2f}
- Model 2 (Profile Classifier): Accuracy={accuracy_m2:.0%}, Silhouette={silhouette:.2f}
- Model 3 (Success Predictor): Accuracy={accuracy_m3:.0%}, AUC={auc_m3:.2f}

🔑 KEY FINDINGS:
- Implementation Paradox: Govt {govt_success:.0f}%, Academic {acad_success:.0f}%, NGO {ngo_success:.0f}%
- {len(df['profile_label'].unique())} distinct readiness profiles identified
- Critical threshold at 75/100 readiness score

📁 All results and visualizations have been downloaded!

🚀 Ready to update your AIES paper with these results!
""")
```
